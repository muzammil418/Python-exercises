## List comprehensions

```python
mylist = []
for i in range(10):
    mylist.append(i)
```

```python
mylist
```

```python
# Syntax:
# [ (VARIABLE OR EXPRESSION) (for LOOP line) (if CONDITIONAL?)]
[tmp for tmp in range(10)]
```

```python
# List comprehension to get odd numbers only

[t for t in range(10) if t % 2 == 1]
```

```python
[t**2 for t in range(10) if t % 2 == 1]
```

```python
# List of even numbers

[t for t in range(10) if t%2==0]
```

```python
def make_my_url(num):
    return f"https://www.example.com/my_api/{num}"

[make_my_url(t) for t in range(10) if t != 5]
```

```python
# Expected output: [ [0, 0], [0, 1], [0, 2].... ]

[ [0, t] for t in range(10) ]
```
