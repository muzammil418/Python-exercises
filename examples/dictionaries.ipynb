```python
raw_data = """Argentina,Buenos Aires
Australia,Canberra
Brazil,Brasília
Canada,Ottawa
China,Beijing
Egypt,Cairo
France,Paris
Germany,Berlin
India,New Delhi
Indonesia,Jakarta
Italy,Rome
Japan,Tokyo
Mexico,Mexico City
Nigeria,Abuja
Russia,Moscow
South Africa,Pretoria
South Korea,Seoul
Spain,Madrid
United Kingdom,London
United States,"Washington, D.C." """
```

```python
raw_data
```

```python
countries_capital_list = raw_data.split('\n')
```

```python
len(countries_capital_list)
```

```python
countries_capital_list
```

```python
# Make dict like: Country -> Capital

countries_dict = dict()

for line in countries_capital_list:
    mykey = line[:line.index(",")]
    myvalue = line[line.index(",")+1:]

    countries_dict[mykey] = myvalue

countries_dict
```

```python
countries_dict_new = {}

for line in countries_capital_list:
    tmp = line.split(',')
    countries_dict_new[tmp[0]] = tmp[1]

countries_dict_new
```

```python
countries_dict_new['Brazil']
```

## Dictionary: keys, values, checking membership using "in" operator

```python
# To get all the keys as list
countries_dict_new.keys()
```

```python
# To get all the dictionary values as list
countries_dict_new.values()
```

```python
mylist = [f"player_{t}" for t in range(10)]

mylist # List of strings
```

```python
"player_11" in mylist
```

```python
# Do we have a capital of Australia in our dictionary mapping???
# "in" operator checks membership in a collection (like list, set, tuple, dictionary, etc)
# In this case, "in" operator already knows that you want to check the membership of a key in the dict (so expilicitly saying mydict.keys()) is not necessary
"Pakistan" in countries_dict_new
```

## Basic dictionary comprehension

```python
# Basic dictionary comprehension
# Basic syntax:
# { KEY_EXPR:VALUE_EXPR FOR_SYNTAX CONDITIONAL?}

{ k:(f"hello_{k}") for k in range(20) }
```

```python
capitla_to_countries_map = {}

for k, v in countries_dict_new.items():
    capitla_to_countries_map[v] = k

capitla_to_countries_map
```

```python
# Using dictionary comprehensions
{v:k for k, v in countries_dict_new.items()}
```
