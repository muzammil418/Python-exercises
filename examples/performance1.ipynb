```python
import time
import random
```

```python
"""
Following dict maps capitals to their countries
This is a multi line comment in python
"""

mydict = {'Buenos Aires': 'Argentina',
 'Canberra': 'Australia',
 'Brasília': 'Brazil',
 'Ottawa': 'Canada',
 'Beijing': 'China',
 'Cairo': 'Egypt',
 'Paris': 'France',
 'Berlin': 'Germany',
 'New Delhi': 'India',
 'Jakarta': 'Indonesia',
 'Rome': 'Italy',
 'Tokyo': 'Japan',
 'Mexico City': 'Mexico',
 'Abuja': 'Nigeria',
 'Moscow': 'Russia',
 'Pretoria': 'South Africa',
 'Seoul': 'South Korea',
 'Madrid': 'Spain',
 'London': 'United Kingdom',
 '"Washington': 'United States'}
```

```python
mydict
```

```python
capitals = list(mydict.keys())
capitals
```

```python
countries = list(mydict.values())
countries
```

```python
million_10 = 10000000
```

```python
# Performance of dict

start = time.perf_counter()

for _ in range(million_10): # One million cycles
    random_key = random.choice(capitals)
    country_of_key = mydict[random_key] # Lookup done here
    # print(f"{random_key} -> {country_of_key}")

elapsed = time.perf_counter() - start
print(f"Elapsed time: {elapsed:.2f} seconds")
```

```python
# Performance of list

start = time.perf_counter()

for _ in range(million_10): # One million cycles
    value_to_search = random.choice(capitals)
    result_i = None
    for i,val in enumerate(capitals): # linear search
        if val == value_to_search:
            result_i = i
            break
    country_of_key = countries[result_i]
    # print(f"{random_key} -> {country_of_key}")

elapsed = time.perf_counter() - start
print(f"Elapsed time: {elapsed:.2f} seconds")
```
